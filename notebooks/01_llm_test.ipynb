```python
groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(
    model="qwen/qwen3.8-27b",
    temperature=0.3,
    api_key=groq_api_key
)
```

```python
response = llm.invoke("What is Generative AI?")

print(response.content)
```

```python
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder
```

```python
store = {}

def get_session_history(session_id:str):
    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()
    return store[session_id]


# Message placeholder is one type of special placeholder to dynamically insert the chathistory or message to the chatprompttemplate at run time
prompt = ChatPromptTemplate.from_messages([
    ('system',"You are a helpful AI assistant. Answer clearly."),
    MessagesPlaceholder(variable_name="chat_history"),
    ('human','{input}')
])
```

```python
chain = prompt | llm


# runnablewithmessagehistory is one type of manager or wrapper to connect the llm with memory. it will automatically fetch the past information and give to the llm .
chatbot = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="input",
    history_messages_key="chat_history"
)
```

```python
response = chatbot.invoke({'input':"My name is Bhabani. I am learning Generative AI."},config={"configurable":{'session_id':'user_1'}})
```
