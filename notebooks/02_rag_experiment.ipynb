```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=900,
    chunk_overlap=150,
    separators=[
        "\n\n",
        "\n",
        ". ",
        " ",
        ""
    ]
)

documents = text_splitter.split_documents(pages)

print("Total chunks:", len(documents))
```

```python
for i, doc in enumerate(documents[:5], start=1):

    print("=" * 80)

    print("CHUNK:", i)

    print("PAGE:", doc.metadata.get("page"))

    print(doc.page_content)
```

```python
from langchain_ollama import OllamaEmbeddings
```

```python
embedding_model = OllamaEmbeddings(
    model="all-minilm")
```

```python
from langchain_community.vectorstores import FAISS
```

```python
vector_store = FAISS.from_documents(
    documents=documents,
    embedding=embedding_model
)
```

```python
query = "What are the main topics explained in this document?"

results = vector_store.similarity_search(
    query,
    k=3
)

for i, doc in enumerate(results, start=1):

    print("=" * 80)

    print("RESULT:", i)

    print(doc.page_content)

    print("METADATA:", doc.metadata)
```

```python
retriever = vector_store.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 3}
)
```

```python
query = "What are the main concepts explained in this document?"

retrieved_docs = retriever.invoke(query)

for i, doc in enumerate(retrieved_docs, start=1):
    print(f"\n--- Document {i} ---")
    print(doc.page_content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

rag_prompt = ChatPromptTemplate.from_template("""
You are a helpful AI assistant.

Answer the user's question using the provided context.

Instructions:
1. Use the context to answer accurately.
2. Explain the answer in simple language.
3. If the answer is not available in the context, clearly say that
   the uploaded document does not contain this information.
4. Do not invent information.

Context:
{context}

User Question:
{question}

Answer:
""")
```

```python
llm = ChatGroq(
    model="qwen/qwen3.8-27b",
    temperature=0.3,
    api_key=api_key
)
```

```python
from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder
from langchain_core.output_parsers import StrOutputParser
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
```

```python
output_parser = StrOutputParser()
```

```python
store = {}

def get_session_history(session_id: str):
    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()
    return store[session_id]

```

```python
contextualize_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """Given the conversation history and the latest user question,
rewrite the latest question as a standalone question.

Rules:
1. Do not answer the question.
2. Use conversation history to understand references such as:
   "it", "its", "they", "this", "that", etc.
3. If the question is already clear, return it unchanged.
4. Return only the standalone question."""
    ),

    MessagesPlaceholder(variable_name="chat_history"),

    ("human", "{question}")
])
```

```python
contextualize_chain = contextualize_prompt | llm | output_parser
```

```python
rag_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """You are a helpful AI assistant.

        Answer using the provided document context.

        If the answer is not available in the document,
        clearly say so.

        Context:
        {context}"""
    ),

    MessagesPlaceholder(variable_name="chat_history"),

    ("human", "{question}")
])
```

```python
rag_chain = rag_prompt | llm | output_parser
```

```python
conversational_rag_chain = RunnableWithMessageHistory(
    rag_chain,
    get_session_history,
    input_messages_key="question",
    history_messages_key="chat_history"
)
```

```python
def ask_question(question,session_id):
    history = get_session_history(session_id)

    if history.messages:
        standalone_question = contextualize_chain.invoke({
            'chat_history':history.messages,
            'question':question
        })
    else:
        standalone_question = question

    print("Original Question:")
    print(question)

    print("\nRetrieval Question:")
    print(standalone_question)

    retrieved_docs = retriever.invoke(standalone_question)

    context = "\n\n".join(
        doc.page_content
        for doc in retrieved_docs
    )

    response = conversational_rag_chain.invoke(
        {
            'context':context,
            'question':question
        },
        config={
            'configurable':{'session_id':session_id}
        }
    )

    return response
```

```python
import uuid
from datetime import datetime
from langchain_core.chat_history import InMemoryChatMessageHistory
```

```python
store = {}
conversations = {}
```

```python
def create_new_chat():

    session_id = str(uuid.uuid4())

    store[session_id] = InMemoryChatMessageHistory()

    conversations[session_id] = {
        "title": "New Chat",
        "created_at": datetime.now()
    }

    return session_id
```

```python
session_id = create_new_chat()

print("Session ID:", session_id)
print("Conversations:", conversations)
print("Messages:", store[session_id].messages)
```

```python
def get_all_conversations():

    chat_list = []

    for session_id, details in conversations.items():

        chat_list.append({
            "session_id": session_id,
            "title": details["title"],
            "created_at": details["created_at"]
        })

    return chat_list
```

```python
def switch_chat(session_id):

    if session_id not in conversations:
        raise ValueError("Conversation not found")

    history = store[session_id]

    return history
```

```python
session_1 = create_new_chat()
session_2 = create_new_chat()
```

```python
all_chats = get_all_conversations()

for chat in all_chats:
    print(chat)
```

```python
def delete_chat(session_id):
    if session_id not in conversations:
        raise ValueError("Conversation not found")
    del conversations[session_id]

    del store[session_id]

    return "Chat deleted sucessfully"
```

```python
def generate_chat_title(question):

    prompt = f"""
    Generate a short title for this conversation.

    User question: {question}

    Rules:
    1. Maximum 5 words.
    2. Do not use quotation marks.
    3. Return only the title.
    """

    response = llm.invoke(prompt)

    return response.content.strip()
```

```python
def update_chat_title(session_id, question):

    if session_id not in conversations:
        raise ValueError("Conversation not found")

    if conversations[session_id]["title"] == "New Chat":

        title = generate_chat_title(question)

        conversations[session_id]["title"] = title

    return conversations[session_id]["title"]
```

```python
session_id = create_new_chat()

title = update_chat_title(
    session_id,
    "What is machine learning and how does it work?"
)

print(title)
```

```python
def ask_question(question, session_id):

    if session_id not in conversations:
        raise ValueError("Conversation not found")

    if conversations[session_id]["title"] == "New Chat":

        update_chat_title(session_id, question)

    # Get conversation history
    history = get_session_history(session_id)

    # Contextualize follow-up questions
    if history.messages:

        standalone_question = contextualize_chain.invoke({
            "chat_history": history.messages,
            "question": question
        })

    else:
        standalone_question = question

    # Retrieve relevant documents
    retrieved_docs = retriever.invoke(standalone_question)

    # Prepare context
    context = "\n\n".join(
        doc.page_content
        for doc in retrieved_docs
    )

    # Generate response
    response = conversational_rag_chain.invoke(
        {
            "context": context,
            "question": question
        },
        config={
            "configurable": {
                "session_id": session_id
            }
        }
    )

    return response
```
